Хранение в памяти O(n)
Каждый проход O(n)
Для n интервалов иза-за сортировки O(nlogn)


```python
from datetime import date


def busiest_day(visits):
    events = []
    for arrival, departure in visits:
        start = date.fromisoformat(arrival).toordinal()
        end = date.fromisoformat(departure).toordinal()

        end += 1
        if start < end:
            events.append((start, 1))
            events.append((end, -1))
    events.sort()
    current = maximum = 0
    best_day = None
    i = 0
    while i < len(events):
        day = events[i][0]
        while i < len(events) and events[i][0] == day:
            current += events[i][1]
            i += 1
        if current > maximum:
            maximum = current
            best_day = day
    if best_day is None:
        return None, 0
    return date.fromordinal(best_day).isoformat(), maximum
```

```python
visits = [('2024-09-15', '2024-09-15'), ('2024-09-14', '2024-09-21')]
```

```python
busiest_day(visits)
```
