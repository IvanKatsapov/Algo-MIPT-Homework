Cложность О(VE + V + E), где V - колиетво вершин,  E - количество ребер, память O(V+E)

```python
from collections import deque


def bellman_ford(n, edges, start):
    graph = [[] for _ in range(n)]
    for u, v, weight in edges:
        graph[u].append(v)

    dists = [float('inf')] * n
    parent = [None] * n
    dists[start] = 0
    for _ in range(n - 1):
        changed = False
        for u, v, weight in edges:
            if dists[u] != float('inf') and dists[u] + weight < dists[v]:
                dists[v] = dists[u] + weight
                parent[v] = u
                changed = True
        if not changed:
            break

    bad = [False] * n
    queue = deque()
    for u, v, weight in edges:
        if dists[u] != float('inf') and dists[u] + weight < dists[v]:
            if not bad[v]:
                bad[v] = True
                queue.append(v)
    while queue:
        u = queue.popleft()
        for v in graph[u]:
            if not bad[v]:
                bad[v] = True
                queue.append(v)
    for v in range(n):
        if bad[v]:
            dists[v] = float('-inf')
            parent[v] = None
    return dists, parent


def restore_path(parent, dists, finish):
    if dists[finish] == float('inf'):
        return None
    if dists[finish] == float('-inf'):
        raise ValueError('Кратчайшего пути нет')
    path = []
    while finish is not None:
        path.append(finish)
        finish = parent[finish]
    path.reverse()
    return path
```
