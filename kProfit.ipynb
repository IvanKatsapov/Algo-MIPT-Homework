Хранение данных O(n) на цены и O(1) на количество сделок


```python
import heapq


def max_profit(prices, k):
    n = len(prices)
    if n < 2 or k == 0:
      return 0
    if k >= n // 2:
      return sum(max(0, prices[i] - prices[i - 1]) for i in range(1, n))

    stack = []
    best = []

    def save(profit):
      if profit <= 0:
        return
      if len(best) < k:
        heapq.heappush(best, profit)
      elif profit > best[0]:
        heapq.heapreplace(best, profit)

    i = 0
    #находим локальные минимумы и максимумы
    while i < n - 1:
        while i < n - 1 and prices[i] >= prices[i + 1]:
            i += 1
        low = prices[i]
        while i < n - 1 and prices[i] <= prices[i + 1]:
            i += 1
        high = prices[i]
        if high == low:
            break
#проверяем участки в стеке, удаляем старые, пока новые ыгоднее
        while stack and low < stack[-1][0]:
            old_low, old_high = stack.pop()
            save(old_high - old_low)
#обобъединяем границы сделок
        while stack and high >= stack[-1][1]:
            old_low, old_high = stack.pop()
            save(old_high - low)
            low = old_low
        stack.append((low, high))

    while stack:
        low, high = stack.pop()
        save(high - low)
    return sum(best)
```

```python
prices = [3, 2, 6, 5, 0, 3]
k = 2
print(max_profit(prices, k))
```
